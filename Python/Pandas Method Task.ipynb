```python
#30
import pandas as pd
df = pd.read_csv('Bank_customer_churn - Bank_customer_churn.csv')
print("Displaying dataset: ",df)
print("check missing values: ",df.isnull().sum())
df_remove = df.drop_duplicates()
df_rename = df_remove.rename(columns={"CustomerId":"Customer_ID","CreditScore":"Credit_Score","Estimated_Salary":"Salary","Exited":"Target"})
df_convert = df_rename.astype({"Age":"float64","Balance":"float64","Target":"string"})
df_convert["Age"] = df_convert["Age"].fillna(df_convert["Age"].median())
df_convert["Balance"] = df_convert["Balance"].fillna(df_convert["Balance"].mean())
df_remove_columns = df_convert.drop(columns=['RowNumber','Surname'])
print("Display first 10 rows: ",df_remove_columns.head(10))
print("Statistical summary: ",df_remove_columns.describe())
print("Value counts of target: ",df_remove_columns["Target"].value_counts())
print("Final DataSet Information: ",df_remove_columns.info())
```
