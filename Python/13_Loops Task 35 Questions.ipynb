```python
#10
num = int(input("Enter the number: "))
total = 0
while num != 0:
    total = total + num
    num = int(input("Enter the number: "))
print("Final Sum:",total)
```

```python
#11
num = int(input("Enter the number: "))
reverse = 0
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num//10
print("Reversed number:",reverse)
```
