```python
#4
import pandas as pd
df = pd.read_csv('EV_Resale_Price_Regression - EV_Resale_Price_Regression.csv')
df["listing_date"] = pd.to_datetime(df["listing_date"],errors = "coerce")
df["listing_year"] = df["listing_date"].dt.year
df["vehicle_age"] = df["listing_year"] - df["manufacture_year"]
print("Vehicle Age:")
print(df[["listing_date","manufacture_year","vehicle_age"]].head(10))
print("\nVehicle with age = 0:")
print(df[df["vehicle_age"]==0])
print("\nVehicles with age < 0:")
print(df[df["vehicle_age"]<0])
print("\nVehicle age statistics:")
print(df["vehicle_age"].describe())
```

```python
#5
import pandas as pd
df = pd.read_csv('EV_Resale_Price_Regression - EV_Resale_Price_Regression.csv')
df["battery_capacity_kwh"] = df["battery_capacity_kwh"].fillna(df["battery_capacity_kwh"].median())
df["battery_health_pct"] = df["battery_health_pct"].fillna(df["battery_health_pct"].median())
df["range_km"] = df["range_km"].fillna(df["range_km"].mean())
df["charging_time_hr"] = df["charging_time_hr"].fillna(df["charging_time_hr"].mean())
print("missing values after filling:")
print(df[
    ["battery_capacity_kwh",
    "battery_health_pct",
    "range_km",
    "charging_time_hr"]
].isnull().sum())
```

```python
#6
import pandas as pd
df = pd.read_csv("EV_Resale_Price_Regression - EV_Resale_Price_Regression.csv")
print("Battery health below 50%:")
print(df[df["battery_health_pct"]<50][["vehicle_id","battery_health_pct"]])
print("\nBattery health above 100%:")
print(df[df["battery_health_pct"]>100][["vehicle_id","battery_health_pct"]])
Q1 = df["battery_health_pct"].quantile(0.25)
Q3 = df["battery_health_pct"].quantile(0.75)
IQR = Q3-Q1
lower_limit = Q1-1.5*IQR
upper_limit = Q3+1.5*IQR
print("\nIQR Lower Limit:",lower_limit)
print("IQR Upper Limit:",upper_limit)
extreme_values = df[
    (df["battery_health_pct"]<lower_limit)|
    (df["battery_health_pct"]>upper_limit)
]
print("\nExtreme battery health observations:")
print(extreme_values[["vehicle_id","battery_health_pct"]])
```

```python
#7
import pandas as pd
df = pd.read_csv('EV_Resale_Price_Regression - EV_Resale_Price_Regression.csv')
df["range_km"] = pd.to_numeric(df["range_km"],errors = "coerce")
df["battery_capacity_kwh"] = pd.to_numeric(df["battery_capacity_kwh"],errors = "coerce")
df["range_per_kwh"] = df["range_km"]/df["battery_capacity_kwh"]
print("Range per kwh:")
print(df[[
    "vehicle_id",
    "range_km",
    "battery_capacity_kwh",
    "range_per_kwh"
]].head(10))
Q1 = df["range_per_kwh"].quantile(0.25)
Q3 = df["range_per_kwh"].quantile(0.75)
IQR = Q3-Q1
lower_limit = Q1-1.5*IQR
upper_limit = Q3+1.5*IQR
print("\nLower efficieny limit:",lower_limit)
print("Higher efficiency limit:",upper_limit)
low_efficiency = df[
    df["range_per_kwh"]<lower_limit
]
print("\nUnusually Low Efficiency:")
print(low_efficiency[[
    "vehicle_id",
    "range_km",
    "battery_capacity_kwh",
    "range_per_kwh"
]])
high_efficiency = df[
    df["range_per_kwh"]>upper_limit
]
print("\nUnusually high efficiency:")
print(high_efficiency[[
    "vehicle_id",
    "range_km",
    "battery_capacity_kwh",
    "range_per_kwh"
]])
```
