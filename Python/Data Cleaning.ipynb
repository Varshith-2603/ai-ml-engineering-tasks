```python
#8
import pandas as pd
import numpy as np
df = pd.read_csv("matplotlib_messy_sales_2000 - matplotlib_messy_sales_2000.csv")
df["Age"] = pd.to_numeric(df["Age"].astype(str).str.extract(r"(\d+)")[0],errors="coerce")
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)
IQR = Q3-Q1
lower_limit = Q1-1.5*IQR
upper_limit = Q3+1.5*IQR
print("Lower Limit:",lower_limit)
print("Upper Limit:",upper_limit)
outliers = df[(df["Age"]<lower_limit)|(df["Age"]>upper_limit)]
print("\nAge Outliers:")
print(outliers["Age"])
df.loc[(df["Age"] < lower_limit) | (df["Age"] > upper_limit), "Age"] = np.nan
print("\nOutliers remaining:",((df["Age"] < lower_limit) | (df["Age"] > upper_limit)).sum())
print("Missing ages after cleaning:",df["Age"].isnull().sum())
```

```python
#19
import pandas as pd
df = pd.read_csv("matplotlib_messy_sales_2000 - matplotlib_messy_sales_2000.csv")
df["Order Date"] = pd.to_datetime(df["Order Date"],errors="coerce")
numerical_columns = ['Age',"Units Sold","Unit Price","Discount %","Customer Rating","Gross Sales","Discount Amount","Revenue","Cost","Profit"]
for col in numerical_columns:
    df[col] = pd.to_numeric(df[col],errors="coerce")
binary_columns = ["Is Active","Returned"]
for col in binary_columns:
    df[col] = pd.to_numeric(df[col],errors="coerce")
    df.loc[~df[col].isin([0, 1]) & df[col].notna(), col] = pd.NA
categorical_columns = ["City", "Product Category", "Customer Segment","Gender", "Sales Channel", "Payment Method"]
for col in categorical_columns:
    df[col] = df[col].astype("string").str.strip()
print("DATAFRAME INFORMATION:")
df.info()
print("\nDate datatype:")
print(df["Order Date"].dtype)
print("\nNumerical column datatypes:")
print(df[numerical_columns].dtypes)
print("\nBinary column values:")
for col in binary_columns:
    print(f"{col}: {df[col].dropna().unique()}")
print("\nCategorical column datatypes:")
print(df[categorical_columns].dtypes)
print("\nRemaining missing values:")
print(df.isnull().sum())
```
