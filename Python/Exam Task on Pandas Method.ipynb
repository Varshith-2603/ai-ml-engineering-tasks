```python
#8
import pandas as pd
df = pd.read_csv("customer_analytics - customer_analytics.csv")
numeric_df = df.select_dtypes(include=['number'])
print("Currently numeric columns:")
print(numeric_df.columns.tolist())
print("\nPotentially Numeric Object Columns:")
for column in df.select_dtypes(include=['object']).columns:
    converted = pd.to_numeric(df[column],errors = 'coerce')
    numeric_values = converted.notna().sum()
    original_values = df[column].notna().sum()
    if numeric_values > 0 and numeric_values < original_values:
        print(column)
```

```python
#9
import pandas as pd
df = pd.read_csv('customer_analytics - customer_analytics.csv')
df['Age'] = pd.to_numeric(df["Age"],errors = 'coerce')
median_age = df['Age'].median()
print("Median age:",median_age)
df['Age'] = df["Age"].fillna(median_age)
print("\nAge after cleaning: ")
print(df["Age"].head(20))
print("/nData Type:")
print(df["Age"].dtype)
print("\nMissing Age values:")
print(df["Age"].isnull().sum())
```

```python
#20
import pandas as pd
df = pd.read_csv('customer_analytics - customer_analytics.csv')
print("Check Frequency:")
print(df["Segment"].value_counts())
print("\nMissing Values:")
print(df["Segment"].isnull().sum())
most_common_segment = df["Segment"].value_counts().idxmax()
print("\nMost common segment:",most_common_segment)
df["Segment"] = df["Segment"].fillna(most_common_segment)
print("\nMissing values after filling:")
print(df["Segment"].isnull().sum())
print("\nFinal segment frequency:")
print(df["Segment"].value_counts())
```
