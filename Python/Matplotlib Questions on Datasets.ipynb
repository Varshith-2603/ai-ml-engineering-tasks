```python
#27
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv("matplotlib_sales_clean_1000 - matplotlib_sales_clean_1000.csv")
columns = ["Customer_Age","Units_Sold","Unit_Price","Discount_Pct","Revenue","Cost","Profit"]
correlation = df[columns].corr()
plt.figure(figsize=(9,7))
plt.imshow(correlation,cmap="coolwarm",vmin=-1,vmax=1)
plt.xticks(range(len(columns)),columns,rotation=45,ha="right")
plt.yticks(range(len(columns)),columns)
plt.colorbar(label="Correlation Coefficient")
for i in range(len(columns)):
    for j in range(len(columns)):
        plt.text(j,i,f"{correlation.iloc[i,j]:.2f}",ha="center",va="center")
plt.title("Correlation heatmap")
plt.tight_layout()
plt.show()
```
