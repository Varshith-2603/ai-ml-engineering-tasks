```python
#20
import pandas as pd
df = pd.read_csv('customer_data - customer_data.csv')
df["income_usd"] = pd.to_numeric(df["income_usd"],errors = "coerce")
Q1 = df["income_usd"].quantile(0.25)
Q3 = df["income_usd"].quantile(0.75)
IQR = Q3-Q1
lower_limit = Q1-1.5*IQR
upper_limit = Q3+1.5*IQR
outliers = df[(df["income_usd"]<lower_limit) | (df["income_usd"]>upper_limit)]
print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)
print("Lower Limit:",lower_limit)
print("Upper Limit:",upper_limit)
print("\nOutliers in income usd:")
display(outliers["income_usd"])
```

```python
#24
import pandas as pd
df = pd.read_csv('customer_data - customer_data.csv')
numeric_columns = df.select_dtypes(include="number").columns
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())
display(df[numeric_columns])
```

```python
#25
import pandas as pd
df = pd.read_csv('customer_data - customer_data.csv')
categorical_columns = df.select_dtypes(include='object').columns
for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode())
display(df[categorical_columns])
```
