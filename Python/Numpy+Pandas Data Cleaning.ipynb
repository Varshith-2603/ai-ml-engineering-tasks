```python
#3
import pandas as pd
df = pd.read_csv('customer_data - customer_data.csv')
print("All column Names:")
print(df.columns.tolist())
print("\nColumns with leading/trailing spaces:")
for column in df.columns:
    if column != column.strip():
        print(repr(column))

print("\nColumn names with inconsistent capitalization:")
for column in df.columns:
    if column != column.lower() and column != column.upper():
        print(repr(column))
```

```python
#10
import pandas as pd
df = pd.read_csv('customer_data - customer_data.csv')
print("Unique values in the customer segment:")
print(df["customer_segment"].unique())
print("\nFrequency of each category:")
print(df["customer_segment"].value_counts(dropna=False))
print("\nValues showing spaces or inconsistent capitalization:")
for value in df["customer_segment"].dropna().unique():
    print(repr(value))
```

```python
#11
import pandas as pd
df = pd.read_csv('customer_data - customer_data.csv')
df["customer_segment"] = (df["customer_segment"].str.strip().str.lower().str.title())
print("Customer segment after cleaning:")
print(df["customer_segment"].unique())
```

```python
#12
import pandas as pd
df = pd.read_csv('customer_data - customer_data.csv')
print("unique values in the gender column:")
print(df["gender"].unique())
df["gender"] = (df["gender"].str.strip().str.lower().replace({"m":"Male","male":"Male","f":"Female","female":"Female"}))
print("\nUnique values after cleaning:")
print(df["gender"].unique())
```

```python
#13
import pandas as pd
df = pd.read_csv('customer_data - customer_data.csv')
df["is_active"] = df["is_active"].replace({"Yes":1,"yes":1,"Y":1,"1":1,"No":0,"no":0,"N":0,"0":0},inplace=True)
print("is active column after cleaning:")
display(df)
```

```python
#14
import pandas as pd
df = pd.read_csv('customer_data - customer_data.csv')
df["churn_flag"] = df["churn_flag"].replace({"Yes":1,"No":0},inplace=True)
print("Dataset after cleaning:")
display(df)
```

```python
#15
import pandas as pd
df = pd.read_csv('customer_data - customer_data.csv')
numeric_income = pd.to_numeric(df["income_usd"],errors = "coerce")
invalid_income = df[
    df["income_usd"].notna() & numeric_income.isna()
]
print("Values in income usd that cannot be interpreted as number:")
print(invalid_income["income_usd"].unique())
```

```python
#16
import pandas as pd
df = pd.read_csv('customer_data - customer_data.csv')
df["income_usd"] = (df["income_usd"].astype(str).str.replace("$","").str.replace(",","").str.strip())
df["income_usd"] = pd.to_numeric(df["income_usd"],errors = "coerce")
print("Cleaned income_usd:")
display(df["income_usd"])
```

```python
#17
import pandas as pd
df = pd.read_csv('customer_data - customer_data.csv')
df["Age"] = df["Age"].astype(str).str.extract(r"(\d+)")[0]
df["Age"] = pd.to_numeric(df["Age"],errors = "coerce")
print("Cleaned Age column:")
display(df["Age"])
```
