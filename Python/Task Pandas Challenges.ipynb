```python
#17
import pandas as pd
df = pd.read_csv('Task_dataset - Task_dataset.csv')
df.drop_duplicates(inplace=True)
df.fillna({
    "Age": df["Age"].median(),
    "AnnualIncome": df["AnnualIncome"].mean(),
    "Spending": df["Spending"].mean(),
    "CreditScore": df["CreditScore"].mean(),
    "Balance": df["Balance"].mean(),
    "Tenure": df["Tenure"].median(),
    "Geography": "Unknown",
    "Gender": "Unknown",
    "Segment": "Unknown"
}, inplace=True)
df.rename(columns={
    "CustomerID": "Customer_ID",
    "Age": "Customer_Age",
    "AnnualIncome": "Annual_Income",
    "Spending": "Customer_Spending",
    "CreditScore": "Credit_Score",
    "Balance": "Account_Balance",
    "Tenure": "Customer_Tenure",
    "Geography": "Country",
    "Gender": "Gender",
    "Segment": "Customer_Segment",
    "Target": "Customer_Status"
}, inplace=True)
df.add_prefix("Company")
print("Statistical Summary: ")
display(df.describe())
print("\nTarget Counts:")
display(df["Customer_Status"].value_counts())
```

```python
#20
import pandas as pd
df = pd.read_csv('Task_dataset - Task_dataset.csv')
df.info()
print("Find Missing Values: ",df.isnull().sum())
df.drop_duplicates(inplace=True)
df.rename(columns={
    "CustomerID": "Customer_ID",
    "AnnualIncome": "Income",
    "CreditScore": "Credit_Score",
    "Target": "CustomerStatus"
}, inplace=True)
df["Age"] = df["Age"].fillna(df["Age"].median().astype(int))
df["Income"] = df["Income"].astype(float)
df["Balance"] = df["Balance"].astype(float)
df["Income"] = df["Income"].fillna(df["Income"].mean())
df["Balance"] = df["Balance"].fillna(df["Balance"].mean())
df.drop(columns=["Segment","Customer_ID"],inplace=True)
df["TotalMoney"] = df["Balance"] + df["Income"]
df["AgeAfter5Years"] = df["Age"] + 5
df["MonthlyIncome"] = df["Income"] / 12
df["CreditScoreDouble"] = df["Credit_Score"] * 2
print("\nFirst 10 Rows:")
display(df.head(10))
print("\nLast 10 Rows:")
display(df.tail(10))
print("\n10 Random Rows:")
display(df.sample(10))
print("\nStatistical Summary:")
display(df.describe())
print("\nTarget Counts:")
display(df["CustomerStatus"].value_counts())
print("\nFinal Dataset Information:")
df.info()
```
