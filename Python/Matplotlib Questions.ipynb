```python
#1
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [10,20,15,25,30]
plt.plot(x,y)
plt.title("Simple line plot")
plt.show()
```

```python
#2
import matplotlib.pyplot as plt
x = [2,4,6,8,10]
y = [3,5,7,9,11]
plt.plot(x,y)
plt.title("Basic line plot with title")
plt.show()
```

```python
#3
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [10,20,15,25,30]
plt.plot(x,y)
plt.title("Basic line plot with labels")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.show()
```

```python
#4
import matplotlib.pyplot as plt
import numpy as np
x = np.array([1,2,3,4,5])
y = 2*x
x1 = [2,4,6,8]
y1 = [3,5,7,9]
plt.plot(x,y,"-.",label='y=2x')
plt.plot(x1,y1,label="Second Line")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Multiple lines on same graph")
plt.legend()
plt.show()
```

```python
#5
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.plot(x,y,color='red',linestyle='-.',marker='o')
plt.title("Basic line plot with line color,line style,marker style")
plt.xlabel("X-Values")
plt.ylabel("Y-Values")
plt.show()
```
