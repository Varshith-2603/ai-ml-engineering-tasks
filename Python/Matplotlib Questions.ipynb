```python
#18
import matplotlib.pyplot as plt
def add_labels(products,revenue):
    for i in range(len(products)):
        plt.text(i,revenue[i],revenue[i],ha='center')
products = ["Product A", "Product B", "Product C", "Product D", "Product E"]
revenue = [450000, 320000, 580000, 275000, 410000]
plt.bar(products,revenue)
add_labels(products,revenue)
plt.xlabel("Products")
plt.ylabel("Revenue")
plt.title("Revenue of 5 different products")
plt.show()
```

```python
#25
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 135, 180, 210, 240]
products = ["A", "B", "C", "D", "E"]
revenue = [450, 320, 580, 275, 410]
experience = [1, 3, 5, 7, 9, 11, 13, 15]
salary = [28000, 36000, 45000, 52000, 68000, 75000, 92000, 105000]
marks = [45, 52, 67, 72, 78, 81, 85, 90, 56, 63, 70, 74, 88, 92, 95, 48, 59, 65, 77, 83]
fig,axes = plt.subplots(2,2,figsize=(12,8))
axes[0][0].plot(months,sales,marker='o')
axes[0][0].set_title("Monthly Sales")
axes[0][0].set_xlabel("Month")
axes[0][0].set_ylabel("Sales")
axes[0][1].bar(products,revenue)
axes[0][1].set_title("Product revenue")
axes[0][1].set_xlabel("Products")
axes[0][1].set_ylabel("Revenue")
axes[1][0].scatter(experience,salary)
axes[1][0].set_title("Experience vs salary")
axes[1][0].set_xlabel("Experience")
axes[1][0].set_ylabel("Salary")
axes[1][1].hist(marks,bins=5,edgecolor='black')
axes[1][1].set_title("Distribution of marks")
axes[1][1].set_xlabel("Marks")
axes[1][1].set_ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
#26
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun","Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
sales = [185, 210, 195, 240, 275, 260, 310, 295, 330, 365, 390, 425]
max_sales = max(sales)
max_index = sales.index(max_sales)
max_month = months[max_index]
plt.figure(figsize=(10,5))
plt.plot(months,sales,marker='o',label="Monthly Sales")
plt.scatter(max_month,max_sales,color='red',s=100)
plt.annotate("Maximum Sales: " + str(max_sales) + "\nMonth: " + max_month,(max_month, max_sales),textcoords="offset points",xytext=(0, 15),ha="center")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly sales chart")
plt.grid(True)
plt.legend()
plt.show()
```

```python
#27
import matplotlib.pyplot as plt
age = [22, 25, 28, 31, 35, 38, 42, 45, 29, 33, 40, 48, 52, 27, 36]
income = [45000, 55000, 72000, 95000, 120000, 85000, 135000, 150000, 68000, 105000, 125000, 90000, 160000, 78000, 115000]
age_normal = []
income_normal =[]
age_high = []
income_high = []
for i in range(len(income)):
    if income[i] > 100000:
        age_high.append(age[i])
        income_high.append(income[i])
    else:
        age_normal.append(age[i])
        income_normal.append(income[i])
plt.scatter(age_normal,income_normal,marker='o',label='Income <= ₹1,00,000')
plt.scatter(age_high,income_high,marker="*",s=150,label="Income > ₹1,00,000")
plt.xlabel("Age")
plt.ylabel("Income (₹)")
plt.title("Age vs Income")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#28
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
online = [120, 150, 180, 160, 200, 220]
offline = [80, 100, 90, 110, 120, 130]
corporate = [50, 70, 60, 80, 90, 100]
plt.bar(months,online,label="Online")
plt.bar(months,offline,label="Offline")
plt.bar(months,corporate,bottom=[online[i]+offline[i] for i in range(len(months))],label="Corporate")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly sales by channel")
plt.legend()
plt.show()
```

```python
#29
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
actual_sales = [185, 210, 195, 240, 260, 250, 310, 280, 330, 350, 375, 410]
target_sales = [200, 200, 210, 230, 270, 260, 300, 310, 320, 360, 390, 400]
plt.plot(months,actual_sales,marker='o',label="Actual Sales")
plt.plot(months,target_sales,marker="s",linestyle="--",label="Target Sales")
for i in range(len(months)):
    if actual_sales[i] < target_sales[i]:
        plt.scatter(months[i],actual_sales[i],color='red',s=100)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Actual sales vs target sales")
plt.legend()
plt.grid(True)
plt.show()
```
