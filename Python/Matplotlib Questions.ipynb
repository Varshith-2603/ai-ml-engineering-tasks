```python
#4
import matplotlib.pyplot as plt
import numpy as np
x = np.array([1,2,3,4,5])
y = 2*x
x1 = [2,4,6,8]
y1 = [3,5,7,9]
plt.plot(x,y,"-.",label='y=2x')
plt.plot(x1,y1,label="Second Line")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Multiple lines on same graph")
plt.legend()
plt.show()
```

```python
#12
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [2,4,6,8,10]
fig,axes = plt.subplots(1,2,figsize=(10,4))
axes[0].plot(x,y,marker='o')
axes[0].set_title("Line Plot")
axes[0].set_xlabel("X")
axes[0].set_ylabel("Y")
axes[1].scatter(x,y)
axes[1].set_title("Scatter Plot")
axes[1].set_xlabel("X")
axes[1].set_ylabel("Y")
plt.tight_layout()
plt.show()
```

```python
#13
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(-10,11)
y = x ** 2
plt.plot(x,y)
plt.title("Graph of y=x**2")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

```python
#14
import matplotlib.pyplot as plt
import numpy as np
x1 = np.array([2,4,6,8,10])
y1 = x1**2
x2 = np.array([1,3,5,7,9])
y2 = 2*x2+5
x3 = np.array([5,10,15,20,25])
y3 = x3**3
plt.plot(x1,y1,marker='o',color='hotpink',label="x1**2")
plt.plot(x2,y2,marker="*",color="indigo",label="2*x2+5")
plt.plot(x3,y3,marker="v",color="red",label="x3**3")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Three mathematical functions")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#15
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
sales = [120, 145, 135, 160, 180, 175,
         210, 225, 200, 240, 260, 285]
fig = plt.figure(figsize=(10,10))
plt.plot(months,sales,marker="o",color="blue",linestyle="-.",label="Montly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.grid(True)
plt.legend()
plt.show()
```

```python
#16
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
sales = [185, 210, 195, 240, 275, 260,
         310, 295, 330, 365, 390, 425]
plt.plot(months,sales,marker="h",color='black',linestyle=":")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.grid(True)
plt.show()
```

```python
#17
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
product_A = [120, 145, 135, 170, 190, 210]
product_B = [100, 130, 150, 140, 175, 200]
plt.plot(months,product_A,marker='o',color='blue',linestyle="-",label="Product A sales for 6 months")
plt.plot(months,product_B,marker='s',color="red",linestyle="--",label="Product B sales for 6 months")
plt.xlabel("Months")
plt.ylabel("Products")
plt.title("Muliple products multi sales every month")
plt.legend()
plt.grid(True)
plt.show()
```
