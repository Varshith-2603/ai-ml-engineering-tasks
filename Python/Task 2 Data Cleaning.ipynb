```python
#30
import pandas as pd
import numpy as np

df = pd.read_csv("messy_customer_loan_dataset - messy_customer_loan_dataset.csv")

# Normalize all column names consistently
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

df = df.drop_duplicates()

# Numeric cleaning
df["age"] = pd.to_numeric(df["age"], errors="coerce")

income = (
    df["annual_income"].astype(str)
    .str.strip()
    .str.replace("INR", "", regex=False)
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
)

lakh = income.str.extract(r"([\d.]+)\s*[lL]")[0]
income.loc[lakh.notna()] = (
    pd.to_numeric(lakh[lakh.notna()], errors="coerce") * 100000
).astype(str)

df["annual_income"] = pd.to_numeric(income, errors="coerce")
df["credit_score"] = pd.to_numeric(df["credit_score"], errors="coerce")

df["loan_amount_inr"] = (
    df["loan_amount_inr"].astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)
df["loan_amount_inr"] = pd.to_numeric(df["loan_amount_inr"], errors="coerce")

df["monthly_visits"] = (
    df["monthly_visits"].astype(str)
    .str.strip()
    .str.lower()
    .replace({"five": "5", "n/a": np.nan, "-": np.nan})
)
df["monthly_visits"] = pd.to_numeric(df["monthly_visits"], errors="coerce")

dti = df["debt_to_income"].astype(str).str.strip()
is_percent = dti.str.endswith("%")
df["debt_to_income"] = pd.to_numeric(
    dti.str.replace("%", "", regex=False)
       .str.replace("x", "", regex=False),
    errors="coerce"
)
df.loc[is_percent, "debt_to_income"] /= 100

# Validate ranges
df.loc[~df["age"].between(0, 100), "age"] = np.nan
df.loc[~df["credit_score"].between(300, 850), "credit_score"] = np.nan
df.loc[~df["debt_to_income"].between(0, 1), "debt_to_income"] = np.nan

# Categorical cleaning
for column in ["employment_type", "city", "loan_type", "loan_approved"]:
    df[column] = df[column].replace(["nan", "n/a", "-", ""], np.nan)

df["employment_type"] = df["employment_type"].str.strip().str.title()

df["city"] = (
    df["city"].str.strip().str.lower()
    .replace({
        "hyderbad": "hyderabad",
        "hyd": "hyderabad",
        "bangalore": "bengaluru"
    })
    .str.title()
)

df["loan_type"] = (
    df["loan_type"].str.strip().str.lower()
    .replace({"personal": "personal loan"})
    .str.title()
)

df["loan_approved"] = (
    df["loan_approved"].str.strip().str.lower()
    .replace({
        "yes": "Yes", "y": "Yes", "1": "Yes",
        "no": "No", "n": "No", "0": "No"
    })
)

# Fill missing values
numeric_columns = [
    "age", "annual_income", "credit_score",
    "loan_amount_inr", "monthly_visits", "debt_to_income"
]

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

categorical_columns = [
    "employment_type", "city", "loan_type", "loan_approved"
]

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

# Replace income outliers with the median
q1 = df["annual_income"].quantile(0.25)
q3 = df["annual_income"].quantile(0.75)
iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

df.loc[
    ~df["annual_income"].between(lower_limit, upper_limit),
    "annual_income"
] = df["annual_income"].median()

print("Remaining missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

display(df)
```
