```python
#12
import pandas as pd
df = pd.read_csv('messy_customer_loan_dataset - messy_customer_loan_dataset.csv')
df["Annual Income"] = (df["Annual Income"].astype(str).str.strip().str.replace("INR","",regex = False).str.replace("₹","",regex = False))
df.loc[df["Annual Income"].str.endswith("L"),"Annual Income"] = (pd.to_numeric(df.loc[df["Annual Income"].str.endswith("L"),"Annual Income"].str.replace("L","",regex = False),errors = "coerce")*100000)
df["Annual Income"] = pd.to_numeric(df["Annual Income"],errors = "coerce")
display(df["Annual Income"])
```

```python
# 12
import pandas as pd
df = pd.read_csv("messy_customer_loan_dataset - messy_customer_loan_dataset.csv")
df["Annual Income"] = (
    df["Annual Income"]
    .astype(str)
    .str.strip()
    .str.replace("INR", "", regex=False)
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
)
df["Annual Income"] = df["Annual Income"].str.replace(
    "L", "00000", regex=False
)
df["Annual Income"] = pd.to_numeric(
    df["Annual Income"],
    errors="coerce"
)
display(df[["Annual Income"]])
```
